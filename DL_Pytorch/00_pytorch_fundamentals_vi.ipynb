## Giới thiệu về tensors 

Bây giờ chúng ta đã nhập PyTorch, đã đến lúc tìm hiểu về tensors.

Tensors là khối xây dựng cơ bản của học máy.

Công việc của chúng là biểu diễn dữ liệu theo cách số học.

Ví dụ, bạn có thể biểu diễn một hình ảnh dưới dạng một tensor có hình dạng (shape) `[3, 224, 224]` nghĩa là `[colour_channels, height, width]`, vì hình ảnh có `3` kênh màu (đỏ, xanh lá, xanh dương), chiều cao (height) là `224` pixel và chiều rộng (width) là `224` pixel.

Trong ngôn ngữ của tensor (ngôn ngữ được sử dụng để mô tả tensors), tensor này sẽ có ba chiều (dimensions), một chiều cho `colour_channels` (kênh màu), `height` (chiều cao) và `width` (chiều rộng).

### Tạo tensors 

PyTorch rất yêu thích tensors. Đến mức có cả một trang tài liệu dành riêng cho lớp [`torch.Tensor`](https://pytorch.org/docs/stable/tensors.html).

Bài tập về nhà đầu tiên của bạn là [đọc qua tài liệu về `torch.Tensor`](https://pytorch.org/docs/stable/tensors.html) trong 10 phút. Nhưng bạn có thể làm điều đó sau.

Hãy code nào.

Điều đầu tiên chúng ta sẽ tạo là một **vô hướng (scalar)**.

Một vô hướng là một số duy nhất và trong ngôn ngữ của tensor, nó là một tensor không chiều (zero dimension).

> **Lưu ý:** Đó là một xu hướng cho khóa học này. Chúng ta sẽ tập trung vào việc viết code cụ thể. Nhưng tôi cũng sẽ thường xuyên giao các bài tập liên quan đến việc đọc và làm quen với tài liệu PyTorch. Bởi vì rốt cuộc, sau khi hoàn thành khóa học này, chắc chắn bạn sẽ muốn học thêm. Và tài liệu là nơi bạn sẽ thường xuyên tìm đến.

````python
# Scalar
scalar = torch.tensor(7)
scalar
````

Thấy cách đoạn code trên in ra `tensor(7)` không?

Điều đó có nghĩa là mặc dù `scalar` là một số duy nhất, nó thuộc loại `torch.Tensor`.

Chúng ta có thể kiểm tra số chiều của một tensor bằng cách sử dụng thuộc tính `ndim`.

````python
scalar.ndim
````

Sẽ thế nào nếu chúng ta muốn lấy số từ tensor ra?

Nói cách khác, chuyển nó từ `torch.Tensor` thành một số nguyên (integer) Python?

Để làm điều này, chúng ta có thể sử dụng phương thức `item()`.

````python
# Get the Python number within a tensor (only works with one-element tensors)
scalar.item()
````

Được rồi, bây giờ chúng ta hãy xem một **vectơ (vector)**.

Một vectơ là một tensor một chiều nhưng có thể chứa nhiều số.

Nói cách khác, bạn có thể có một vectơ `[3, 2]` để mô tả `[số phòng ngủ, số phòng tắm]` trong ngôi nhà của bạn. Hoặc bạn có thể có `[3, 2, 2]` để mô tả `[số phòng ngủ, số phòng tắm, bãi đỗ xe]` trong ngôi nhà của bạn.

Xu hướng quan trọng ở đây là một vectơ rất linh hoạt trong những gì nó có thể biểu diễn (tương tự với tensors).

````python
# Vector
vector = torch.tensor([7, 7])
vector
````

Tuyệt vời, `vector` bây giờ chứa hai số 7, con số yêu thích của tôi.

Bạn nghĩ nó sẽ có bao nhiêu chiều?

````python
# Check the number of dimensions of vector
vector.ndim
````

Hmm, thật kỳ lạ, `vector` chứa hai số nhưng chỉ có một chiều duy nhất.

Tôi sẽ bật mí cho bạn một mẹo.

Bạn có thể biết số chiều của một tensor trong PyTorch bằng số lượng dấu ngoặc vuông ở bên ngoài (`[`) và bạn chỉ cần đếm một bên.

`vector` có bao nhiêu dấu ngoặc vuông?

Một khái niệm quan trọng khác đối với tensors là thuộc tính `shape` (hình dạng) của chúng. Hình dạng cho bạn biết cách các phần tử bên trong chúng được sắp xếp.

Hãy kiểm tra hình dạng của `vector`.

````python
# Check shape of vector
vector.shape
````

Đoạn code trên trả về `torch.Size([2])` có nghĩa là vectơ của chúng ta có hình dạng là `[2]`. Điều này là do hai phần tử chúng ta đã đặt bên trong dấu ngoặc vuông (`[7, 7]`).

Bây giờ chúng ta hãy xem một **ma trận (matrix)**.

````python
# Matrix
MATRIX = torch.tensor([[7, 8], 
                       [9, 10]])
MATRIX
````

Chà! Nhiều số hơn! Các ma trận cũng linh hoạt như các vectơ, ngoại trừ việc chúng có thêm một chiều.

````python
# Check number of dimensions
MATRIX.ndim
````

`MATRIX` có hai chiều (bạn có đếm số dấu ngoặc vuông ở một bên không?).

Bạn nghĩ nó sẽ có `shape` (hình dạng) như thế nào?

````python
MATRIX.shape
````

Chúng ta nhận được đầu ra `torch.Size([2, 2])` vì `MATRIX` có chiều sâu là 2 phần tử và chiều rộng là 2 phần tử.

Còn việc chúng ta tạo một **tensor** thì sao?

````python
# Tensor
TENSOR = torch.tensor([[[1, 2, 3],
                        [3, 6, 9],
                        [2, 4, 5]]])
TENSOR
````

Wow! Thật là một tensor đẹp mắt.

Tôi muốn nhấn mạnh rằng các tensor có thể biểu diễn hầu như mọi thứ.

Tensor chúng ta vừa tạo có thể là doanh số bán hàng của một cửa hàng bít tết và bơ hạnh nhân (hai trong số những món ăn yêu thích của tôi).

Và hình dạng của nó thì sao?

````python
# Check shape of TENSOR
TENSOR.shape
````

### Random tensors (Tensors ngẫu nhiên)

Chúng ta đã xác định rằng các tensors đại diện cho một số dạng dữ liệu.

Và các mô hình học máy, chẳng hạn như mạng nơ-ron (neural networks), sẽ thao tác và tìm kiếm các khuôn mẫu bên trong các tensors.

Tuy nhiên, khi xây dựng các mô hình học máy với PyTorch, rất hiếm khi bạn tạo các tensors thủ công (như những gì chúng ta vừa làm).

Thay vào đó, một mô hình học máy thường bắt đầu với các tensors chứa các con số ngẫu nhiên rất lớn, sau đó điều chỉnh các con số này khi nó xử lý dữ liệu để biểu diễn dữ liệu tốt hơn.

Về cơ bản:

`Bắt đầu với các số ngẫu nhiên -> xem dữ liệu -> cập nhật các số ngẫu nhiên -> xem dữ liệu -> cập nhật các số ngẫu nhiên...`

Là một nhà khoa học dữ liệu, bạn có thể xác định cách mô hình học máy bắt đầu (khởi tạo - initialization), xem xét dữ liệu (biểu diễn - representation) và cập nhật (tối ưu hóa - optimization) các số ngẫu nhiên của nó.

Chúng ta sẽ thực hành các bước này sau.

Hiện tại, hãy xem cách tạo một tensor với các số ngẫu nhiên.

Chúng ta có thể làm điều này bằng cách sử dụng [`torch.rand()`](https://pytorch.org/docs/stable/generated/torch.rand.html) và truyền vào tham số `size`.

````python
# Create a random tensor of size (3, 4)
random_tensor = torch.rand(size=(3, 4))
random_tensor, random_tensor.dtype
````

Tính linh hoạt của `torch.rand()` nằm ở chỗ chúng ta có thể điều chỉnh `size` thành bất cứ thứ gì chúng ta muốn.

Ví dụ, giả sử bạn muốn một tensor ngẫu nhiên với hình dạng phổ biến của hình ảnh là `[224, 224, 3]` (`[chiều_cao, chiều_rộng, kênh_màu]`).

````python
# Create a random tensor of size (224, 224, 3)
random_image_size_tensor = torch.rand(size=(224, 224, 3))
random_image_size_tensor.shape, random_image_size_tensor.ndim
````

### Zeros and ones (Toàn số 0 và toàn số 1)

Đôi khi bạn sẽ chỉ muốn điền vào tensors các số 0 hoặc số 1.

Điều này thường xảy ra khi che giấu (masking - như việc che một số giá trị trong tensor bằng số không để mô hình biết rằng không cần học chúng).

### Tạo một phạm vi (range) và các tensors tương tự (tensors like)

### Các kiểu dữ liệu của tensor

Có rất nhiều [kiểu dữ liệu tensor khác nhau có sẵn trong PyTorch](https://pytorch.org/docs/stable/tensors.html#data-types).

Một số kiểu dành riêng cho CPU và một số kiểu tốt hơn cho GPU.

Việc làm quen để biết cái nào là cái nào có thể mất một chút thời gian.

Nói chung, nếu bạn thấy `torch.cuda` ở bất kỳ đâu, tensor đó đang được sử dụng cho GPU (vì các GPU của Nvidia sử dụng một bộ công cụ tính toán gọi là CUDA).

Kiểu phổ biến nhất (và nói chung là mặc định) là `torch.float32` hoặc `torch.float`.

Được gọi là "số thực dấu phẩy động 32-bit" (32-bit floating point).

Nhưng cũng có số thực dấu phẩy động 16-bit (`torch.float16` hoặc `torch.half`) và số thực dấu phẩy động 64-bit (`torch.float64` hoặc `torch.double`).

Và để làm mọi thứ rối thêm một chút, còn có các số nguyên (integers) 8-bit, 16-bit, 32-bit và 64-bit.

Cộng thêm nhiều thứ khác nữa!

> **Lưu ý:** Số nguyên (integer) là một số tròn phẳng như `7`, trong khi số thực (float) có phần thập phân `7.0`.

Lý do cho tất cả những điều này liên quan đến **độ chính xác trong điện toán (precision in computing)**.

Độ chính xác là mức độ chi tiết được sử dụng để mô tả một con số.

Giá trị độ chính xác càng cao (8, 16, 32), thì càng có nhiều chi tiết và do đó càng tốn nhiều dữ liệu để biểu diễn một con số.

Điều này rất quan trọng trong học sâu và tính toán số học vì bạn đang thực hiện rất nhiều phép toán; bạn càng có nhiều chi tiết để tính toán, bạn càng phải sử dụng nhiều tài nguyên tính toán.

Do đó, các kiểu dữ liệu có độ chính xác thấp hơn thường tính toán nhanh hơn nhưng phải hy sinh một số hiệu suất trên các chỉ số đánh giá như độ chính xác (tính toán nhanh hơn nhưng ít chính xác hơn).

> **Tài nguyên:** 
  * Xem [tài liệu PyTorch để biết danh sách tất cả các kiểu dữ liệu tensor có sẵn](https://pytorch.org/docs/stable/tensors.html#data-types).
  * Đọc [trang Wikipedia để có cái nhìn tổng quan về độ chính xác trong điện toán là gì](https://en.wikipedia.org/wiki/Precision_(computer_science)).

Hãy xem cách tạo một số tensors với các kiểu dữ liệu cụ thể. Chúng ta có thể làm như vậy bằng cách sử dụng tham số `dtype`.

````python
# Default datatype for tensors is float32
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # defaults to None, which is torch.float32 or whatever datatype is passed
                               device=None, # defaults to None, which uses the default tensor type
                               requires_grad=False) # if True, operations performed on the tensor are recorded 

float_32_tensor.shape, float_32_tensor.dtype, float_32_tensor.device
````

### Nhân ma trận (tất cả những gì bạn cần)

Một trong những phép toán phổ biến nhất trong các thuật toán học máy và học sâu (như mạng nơ-ron) là [nhân ma trận](https://www.mathsisfun.com/algebra/matrix-multiplying.html).

PyTorch triển khai chức năng nhân ma trận trong phương thức [`torch.matmul()`](https://pytorch.org/docs/stable/generated/torch.matmul.html).

Hai quy tắc chính đối với phép nhân ma trận cần nhớ là:

1. Các **chiều bên trong (inner dimensions)** phải khớp nhau:
  * `(3, 2) @ (3, 2)` sẽ không hoạt động
  * `(2, 3) @ (3, 2)` sẽ hoạt động
  * `(3, 2) @ (2, 3)` sẽ hoạt động
2. Ma trận kết quả có hình dạng của các **chiều bên ngoài (outer dimensions)**:
 * `(2, 3) @ (3, 2)` -> `(2, 2)`
 * `(3, 2) @ (2, 3)` -> `(3, 3)`

> **Lưu ý:** "`@`" trong Python là ký hiệu cho phép nhân ma trận.

> **Tài nguyên:** Bạn có thể xem tất cả các quy tắc cho phép nhân ma trận bằng `torch.matmul()` [trong tài liệu PyTorch](https://pytorch.org/docs/stable/generated/torch.matmul.html).

Hãy tạo một tensor và thực hiện phép nhân từng phần tử cùng với phép nhân ma trận trên nó.

````python
import torch
tensor = torch.tensor([1, 2, 3])
tensor.shape
````

Sự khác biệt giữa nhân từng phần tử và nhân ma trận là việc cộng dồn các giá trị.

Đối với biến `tensor` của chúng ta có các giá trị `[1, 2, 3]`:

| Phép toán | Tính toán | Code |
| ----- | ----- | ----- |
| **Nhân từng phần tử (Element-wise multiplication)** | `[1*1, 2*2, 3*3]` = `[1, 4, 9]` | `tensor * tensor` |
| **Nhân ma trận (Matrix multiplication)** | `[1*1 + 2*2 + 3*3]` = `[14]` | `tensor.matmul(tensor)` |

````python
# Element-wise matrix multiplication
tensor * tensor
````

````python
# Matrix multiplication
torch.matmul(tensor, tensor)
````

````python
# Can also use the "@" symbol for matrix multiplication, though not recommended
tensor @ tensor
````

Bạn có thể thực hiện nhân ma trận bằng tay nhưng không nên.

Phương thức tích hợp `torch.matmul()` nhanh hơn nhiều.

````python
%%time
# Matrix multiplication by hand 
# (avoid doing operations with for loops at all cost, they are computationally expensive)
value = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]
value
````

````python
%%time
torch.matmul(tensor, tensor)
````

## Một trong những lỗi phổ biến nhất trong học sâu (lỗi hình dạng - shape errors)

Bởi vì phần lớn quá trình học sâu là nhân và thực hiện các phép toán trên ma trận và ma trận có một quy tắc nghiêm ngặt về hình dạng và kích thước nào có thể kết hợp với nhau, một trong những lỗi phổ biến nhất mà bạn sẽ gặp phải trong học sâu là hình dạng không khớp (shape mismatches).

````python
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B) # (this will error)
````

Chúng ta có thể làm cho phép nhân ma trận hoạt động giữa `tensor_A` và `tensor_B` bằng cách làm cho các chiều bên trong (inner dimensions) của chúng khớp nhau.

Một trong những cách để làm điều này là sử dụng phép **chuyển vị (transpose)** (hoán đổi các chiều của một tensor cho trước).

Bạn có thể thực hiện chuyển vị trong PyTorch bằng một trong hai cách:
* `torch.transpose(input, dim0, dim1)` - trong đó `input` là tensor mong muốn cần chuyển vị và `dim0`, `dim1` là các chiều sẽ được hoán đổi.
* `tensor.T` - trong đó `tensor` là tensor mong muốn cần chuyển vị.

Hãy thử cách thứ hai.

````python
# View tensor_A and tensor_B
print(tensor_A)
print(tensor_B)
````

````python
# View tensor_A and tensor_B.T
print(tensor_A)
print(tensor_B.T)
````

````python
# The operation works when tensor_B is transposed
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}\n")
print(f"New shapes: tensor_A = {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}\n")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} <- inner dimensions match\n")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output) 
print(f"\nOutput shape: {output.shape}")
````

Bạn cũng có thể sử dụng [`torch.mm()`](https://pytorch.org/docs/stable/generated/torch.mm.html), đây là dạng viết tắt của `torch.matmul()`.

````python
# torch.mm is a shortcut for matmul
torch.mm(tensor_A, tensor_B.T)
````

Nếu không có phép chuyển vị, các quy tắc của phép nhân ma trận không được thỏa mãn và chúng ta sẽ gặp lỗi như trên.

Cần một hình ảnh trực quan? 

![visual demo of matrix multiplication](https://github.com/mrdbourke/pytorch-deep-learning/raw/main/images/00-matrix-multiply-crop.gif)

Bạn có thể tạo các hình ảnh trực quan cho phép nhân ma trận của riêng mình như thế này tại http://matrixmultiplication.xyz/.

> **Lưu ý:** Một phép nhân ma trận như thế này cũng được gọi là [**tích vô hướng (dot product)**](https://www.mathsisfun.com/algebra/vectors-dot-product.html) của hai ma trận.

Các mạng nơ-ron chứa đầy các phép nhân ma trận và tích vô hướng.

Mô-đun [`torch.nn.Linear()`](https://pytorch.org/docs/1.9.1/generated/torch.nn.Linear.html) (chúng ta sẽ thấy mô-đun này hoạt động ở phần sau), còn được gọi là lớp tiến thẳng (feed-forward layer) hay lớp kết nối đầy đủ (fully connected layer), triển khai một phép nhân ma trận giữa một đầu vào `x` và một ma trận trọng số `A`.

$$
y = x\cdot{A^T} + b
$$

Trong đó:
* `x` là đầu vào của lớp (học sâu là một ngăn xếp của các lớp như `torch.nn.Linear()` và các lớp khác xếp chồng lên nhau).
* `A` là ma trận trọng số (weights matrix) được tạo bởi lớp, bắt đầu là các số ngẫu nhiên sẽ được điều chỉnh khi mạng nơ-ron học cách biểu diễn tốt hơn các khuôn mẫu trong dữ liệu (lưu ý chữ "`T`", đó là vì ma trận trọng số được chuyển vị).
  * **Lưu ý:** Bạn cũng có thể thường xuyên thấy `W` hoặc một chữ cái khác như `X` được sử dụng để chỉ ma trận trọng số.
* `b` là hạng tử thiên vị (bias term) được sử dụng để bù trừ nhẹ cho các trọng số và đầu vào.
* `y` là đầu ra (sự thao tác của đầu vào với hy vọng khám phá ra các khuôn mẫu trong đó).

Đây là một hàm tuyến tính (linear function - bạn có thể đã thấy một thứ tương tự như $y = mx+b$ ở trường trung học hoặc nơi khác), và có thể được sử dụng để vẽ một đường thẳng!

Hãy chơi đùa một chút với lớp tuyến tính.

Thử thay đổi các giá trị của `in_features` và `out_features` bên dưới và xem điều gì xảy ra.

Bạn có nhận thấy điều gì liên quan đến hình dạng không?

````python
# Since the linear layer starts with a random weights matrix, let's make it reproducible (more on this later)
torch.manual_seed(42)
# This uses matrix multiplication
linear = torch.nn.Linear(in_features=2, # in_features = matches inner dimension of input 
                         out_features=6) # out_features = describes outer value 
x = tensor_A
output = linear(x)
print(f"Input shape: {x.shape}\n")
print(f"Output:\n{output}\n\nOutput shape: {output.shape}")
````

> **Câu hỏi:** Điều gì xảy ra nếu bạn thay đổi `in_features` từ 2 thành 3 ở trên? Nó có báo lỗi không? Bạn có thể thay đổi hình dạng của đầu vào (`x`) như thế nào để xử lý lỗi đó? Gợi ý: chúng ta đã phải làm gì với `tensor_B` ở trên?

### Định hình lại (Reshaping), xếp chồng (stacking), ép (squeezing) và giãn (unsqueezing)

Thường thì bạn sẽ muốn định hình lại (reshape) hoặc thay đổi số chiều của tensor mà không thực sự thay đổi các giá trị bên trong chúng.

Để làm như vậy, một số phương thức phổ biến là:

| Phương thức | Mô tả ngắn gọn |
| ----- | ----- |
| [`torch.reshape(input, shape)`](https://pytorch.org/docs/stable/generated/torch.reshape.html#torch.reshape) | Định hình lại `input` theo `shape` (nếu tương thích), cũng có thể sử dụng `torch.Tensor.reshape()`. |
| [`Tensor.view(shape)`](https://pytorch.org/docs/stable/generated/torch.Tensor.view.html) | Trả về một góc nhìn (view) của tensor gốc theo một `shape` khác nhưng chia sẻ cùng một dữ liệu với tensor gốc. |
| [`torch.stack(tensors, dim=0)`](https://pytorch.org/docs/1.9.1/generated/torch.stack.html) | Nối một chuỗi các `tensors` dọc theo một chiều mới (`dim`), tất cả `tensors` phải có cùng kích thước. |
| [`torch.squeeze(input)`](https://pytorch.org/docs/stable/generated/torch.squeeze.html) | Ép `input` để loại bỏ tất cả các chiều có giá trị là `1`. |
| [`torch.unsqueeze(input, dim)`](https://pytorch.org/docs/1.9.1/generated/torch.unsqueeze.html) | Trả về `input` với một chiều có giá trị là `1` được thêm vào tại `dim`. | 
| [`torch.permute(input, dims)`](https://pytorch.org/docs/stable/generated/torch.permute.html) | Trả về một *góc nhìn (view)* của `input` ban đầu với các chiều của nó được hoán vị (sắp xếp lại) theo `dims`. | 

Tại sao lại làm bất kỳ điều nào trong số này?

Bởi vì các mô hình học sâu (mạng nơ-ron) hoàn toàn xoay quanh việc thao tác các tensor theo cách nào đó. Và do các quy tắc của phép nhân ma trận, nếu bạn gặp phải sự không khớp về hình dạng, bạn sẽ gặp lỗi. Các phương thức này giúp bạn đảm bảo rằng các phần tử thích hợp của tensor đang kết hợp với các phần tử thích hợp của các tensor khác. 

Hãy thử nghiệm chúng.

Đầu tiên, chúng ta sẽ tạo một tensor.

````python
# Create a tensor
import torch
x = torch.arange(1., 8.)
x, x.shape
````

Bây giờ hãy thêm một chiều nữa bằng `torch.reshape()`.

````python
# Add an extra dimension
x_reshaped = x.reshape(1, 7)
x_reshaped, x_reshaped.shape
````

Chúng ta cũng có thể thay đổi góc nhìn với `torch.view()`.

````python
# Change view (keeps same data as original but changes view)
# See more: https://stackoverflow.com/a/54507446/7900723
z = x.view(1, 7)
z, z.shape
````

Tuy nhiên, hãy nhớ rằng việc thay đổi góc nhìn của một tensor bằng `torch.view()` thực sự chỉ tạo ra một góc nhìn mới của *cùng* một tensor.

Vì vậy, việc thay đổi góc nhìn cũng làm thay đổi luôn tensor ban đầu.

````python
# Changing z changes x
z[:, 0] = 5
z, x
````

Nếu chúng ta muốn xếp chồng tensor mới của mình lên chính nó năm lần, chúng ta có thể làm như vậy với `torch.stack()`.

````python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0) # try changing dim to dim=1 and see what happens
x_stacked
````

Còn việc xóa tất cả các chiều có giá trị là 1 khỏi một tensor thì sao?

Để làm điều đó, bạn có thể sử dụng `torch.squeeze()` (tôi nhớ điều này như việc *ép* (squeeze) tensor lại để nó chỉ còn các chiều lớn hơn 1).

````python
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
````

Và để thực hiện ngược lại với `torch.squeeze()`, bạn có thể sử dụng `torch.unsqueeze()` để thêm một chiều có giá trị là 1 vào một chỉ mục cụ thể.

````python
print(f"Previous tensor: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
````

Bạn cũng có thể sắp xếp lại thứ tự của các trục bằng `torch.permute(input, dims)`, trong đó `input` được chuyển thành một *góc nhìn (view)* với các `dims` mới.

````python
# Create tensor with specific shape
x_original = torch.rand(size=(224, 224, 3))

# Permute the original tensor to rearrange the axis order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
````

> **Lưu ý**: Bởi vì việc hoán vị trả về một *góc nhìn (view)* (chia sẻ cùng dữ liệu với tensor gốc), các giá trị trong tensor được hoán vị sẽ giống với tensor gốc và nếu bạn thay đổi các giá trị trong góc nhìn, nó sẽ làm thay đổi các giá trị của tensor gốc.

Hai phương thức chính bạn sẽ muốn sử dụng để chuyển từ NumPy sang PyTorch (và ngược lại) là:

````python
# NumPy array to tensor
import torch
import numpy as np
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
array, tensor
````

> **Lưu ý:** Theo mặc định, các mảng NumPy được tạo với kiểu dữ liệu `float64` và nếu bạn chuyển đổi nó thành một tensor PyTorch, nó sẽ giữ nguyên kiểu dữ liệu (như ở trên).

>

Bởi vì chúng ta đã gán lại `tensor` ở trên, nếu bạn thay đổi tensor, mảng vẫn giữ nguyên.

````python
# Change the array, keep the tensor
array = array + 1
array, tensor
````

Và nếu bạn muốn chuyển từ tensor PyTorch sang mảng NumPy, bạn có thể gọi `tensor.numpy()`.

````python
# Tensor to NumPy array
tensor = torch.ones(7) # create a tensor of ones with dtype=float32
numpy_tensor = tensor.numpy() # will be dtype=float32 unless changed
tensor, numpy_tensor
````

Và quy tắc tương tự như trên cũng được áp dụng, nếu bạn thay đổi `tensor` gốc, `numpy_tensor` mới vẫn giữ nguyên.

````python
# Change the tensor, keep the array the same
tensor = tensor + 1
tensor, numpy_tensor
````

## Tính tái lập (cố gắng loại bỏ tính ngẫu nhiên)

Khi bạn tìm hiểu thêm về mạng nơ-ron và học máy, bạn sẽ bắt đầu khám phá ra mức độ ảnh hưởng của sự ngẫu nhiên.

À, đó là sự giả ngẫu nhiên (pseudorandomness). Bởi vì suy cho cùng, theo thiết kế, một máy tính về cơ bản là mang tính quyết định (deterministic - mỗi bước đều có thể dự đoán được) nên sự ngẫu nhiên mà chúng tạo ra là sự ngẫu nhiên được mô phỏng (mặc dù cũng có tranh luận về điều này, nhưng vì tôi không phải là một nhà khoa học máy tính, tôi sẽ để bạn tự tìm hiểu thêm).

Vậy điều này liên quan thế nào đến mạng nơ-ron và học sâu?

Chúng ta đã thảo luận rằng các mạng nơ-ron bắt đầu với các số ngẫu nhiên để mô tả các khuôn mẫu trong dữ liệu (những con số này là những mô tả kém) và cố gắng cải thiện những con số ngẫu nhiên đó bằng cách sử dụng các phép toán tensor (và một vài thứ khác mà chúng ta chưa thảo luận) để mô tả tốt hơn các khuôn mẫu trong dữ liệu.

Tóm lại: 

`bắt đầu với các số ngẫu nhiên -> các phép toán tensor -> cố gắng làm cho tốt hơn (lặp đi lặp lại)`

Mặc dù sự ngẫu nhiên rất tốt và mạnh mẽ, nhưng đôi khi bạn muốn bớt đi một chút sự ngẫu nhiên.

Tại sao?

Để bạn có thể thực hiện các thí nghiệm có thể lặp lại (repeatable experiments).

Ví dụ, bạn tạo ra một thuật toán có khả năng đạt được hiệu suất X.

Và sau đó bạn của bạn dùng thử nó để xác minh rằng bạn không hề điên.

Làm thế nào họ có thể làm được như vậy?

Đó là lúc **tính tái lập (reproducibility)** phát huy tác dụng.

Nói cách khác, bạn có thể nhận được kết quả tương tự (hoặc rất giống) trên máy tính của mình khi chạy cùng một đoạn mã như tôi nhận được trên máy tính của tôi không?

Hãy xem một ví dụ ngắn về tính tái lập trong PyTorch.

Chúng ta sẽ bắt đầu bằng cách tạo hai tensors ngẫu nhiên, vì chúng ngẫu nhiên, bạn sẽ hy vọng chúng khác nhau đúng không?

````python
import torch

# Create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A:\n{random_tensor_A}\n")
print(f"Tensor B:\n{random_tensor_B}\n")
print(f"Does Tensor A equal Tensor B? (anywhere)")
random_tensor_A == random_tensor_B
````

Đúng như bạn có thể mong đợi, các tensors ra kết quả với các giá trị khác nhau.

Nhưng nếu bạn muốn tạo hai tensors ngẫu nhiên với *cùng* giá trị thì sao.

Nghĩa là, các tensors vẫn chứa các giá trị ngẫu nhiên nhưng chúng sẽ có cùng một hương vị (flavour).

Đó là lúc [`torch.manual_seed(seed)`](https://pytorch.org/docs/stable/generated/torch.manual_seed.html) xuất hiện, trong đó `seed` là một số nguyên (như `42` nhưng có thể là bất cứ số nào) để tạo ra "hương vị" cho sự ngẫu nhiên.

Hãy thử bằng cách tạo một số tensors ngẫu nhiên *có hương vị*.

````python
import torch
import random

# # Set the random seed
RANDOM_SEED=42 # try changing this to different values and see what happens to the numbers below
torch.manual_seed(seed=RANDOM_SEED) 
random_tensor_C = torch.rand(3, 4)

# Have to reset the seed every time a new rand() is called 
# Without this, tensor_D would be different to tensor_C 
torch.random.manual_seed(seed=RANDOM_SEED) # try commenting this line out and seeing what happens
random_tensor_D = torch.rand(3, 4)

print(f"Tensor C:\n{random_tensor_C}\n")
print(f"Tensor D:\n{random_tensor_D}\n")
print(f"Does Tensor C equal Tensor D? (anywhere)")
random_tensor_C == random_tensor_D
````

Tuyệt quá!

Có vẻ như việc thiết lập seed đã hoạt động. 

> **Tài nguyên:** Những gì chúng ta vừa đề cập chỉ là bề nổi của tính tái lập trong PyTorch. Để biết thêm về tính tái lập nói chung và các random seeds, tôi khuyên bạn nên xem:
> * [Tài liệu về tính tái lập của PyTorch](https://pytorch.org/docs/stable/notes/randomness.html) (một bài tập tốt là đọc lướt qua phần này trong 10 phút, và ngay cả khi bạn không hiểu nó bây giờ, việc nhận thức được nó vẫn rất quan trọng).
> * [Trang Wikipedia về random seed](https://en.wikipedia.org/wiki/Random_seed) (điều này sẽ cung cấp một cái nhìn tổng quan tốt về các random seeds và sự giả ngẫu nhiên nói chung).

### 2. Thiết lập để PyTorch chạy trên GPU

Khi bạn đã có một GPU sẵn sàng để truy cập, bước tiếp theo là thiết lập PyTorch sử dụng nó để lưu trữ dữ liệu (tensors) và tính toán dữ liệu (thực hiện các phép toán trên tensors).

Để làm vậy, bạn có thể sử dụng gói (package) [`torch.cuda`](https://pytorch.org/docs/stable/cuda.html).

Thay vì nói về nó, chúng ta hãy thử xem sao.

Bạn có thể kiểm tra xem PyTorch có quyền truy cập vào GPU hay không bằng cách sử dụng [`torch.cuda.is_available()`](https://pytorch.org/docs/stable/generated/torch.cuda.is_available.html#torch.cuda.is_available).

````python
# Check for GPU
import torch
torch.cuda.is_available()
````

Nếu kết quả ở trên là `True`, PyTorch có thể thấy và sử dụng GPU, nếu nó trả về `False`, nó không thể thấy GPU và trong trường hợp đó, bạn sẽ phải làm lại các bước cài đặt.

Bây giờ, giả sử bạn muốn thiết lập đoạn mã của mình để nó chạy trên CPU *hoặc* trên GPU nếu có sẵn.

Bằng cách đó, nếu bạn hoặc ai đó quyết định chạy đoạn mã của bạn, nó sẽ hoạt động bất kể thiết bị tính toán mà họ đang sử dụng là gì. 

Hãy tạo một biến `device` để lưu trữ loại thiết bị nào đang khả dụng.

````python
# Set device type
device = "cuda" if torch.cuda.is_available() else "cpu"
device
````

Nếu kết quả ở trên là `"cuda"`, điều đó có nghĩa là chúng ta có thể thiết lập toàn bộ mã PyTorch của mình sử dụng thiết bị CUDA khả dụng (một GPU) và nếu nó trả về `"cpu"`, mã PyTorch của chúng ta sẽ sử dụng CPU.

> **Lưu ý:** Trong PyTorch, phương pháp tốt nhất là viết [**mã không phụ thuộc thiết bị (device agnostic code)**](https://pytorch.org/docs/master/notes/cuda.html#device-agnostic-code). Điều này có nghĩa là mã sẽ chạy trên CPU (luôn có sẵn) hoặc GPU (nếu có).

Nếu bạn muốn thực hiện tính toán nhanh hơn, bạn có thể sử dụng một GPU, nhưng nếu bạn muốn tính toán *nhanh hơn rất nhiều*, bạn có thể sử dụng nhiều GPU.

Bạn có thể đếm số lượng GPU mà PyTorch có quyền truy cập bằng cách sử dụng [`torch.cuda.device_count()`](https://pytorch.org/docs/stable/generated/torch.cuda.device_count.html#torch.cuda.device_count).

````python
# Count number of devices
torch.cuda.device_count()
````

Biết được số lượng GPU mà PyTorch có thể truy cập rất hữu ích trong trường hợp bạn muốn chạy một quá trình cụ thể trên một GPU và một quá trình khác trên một GPU khác (PyTorch cũng có các tính năng cho phép bạn chạy một quá trình trên *tất cả* các GPU).

### 2.1 Chạy PyTorch trên Apple Silicon

Để chạy PyTorch trên các GPU M1/M2/M3 của Apple, bạn có thể sử dụng mô-đun [`torch.backends.mps`](https://pytorch.org/docs/stable/notes/mps.html).

Đảm bảo rằng các phiên bản macOS và PyTorch đã được cập nhật.

Bạn có thể kiểm tra xem PyTorch có quyền truy cập vào GPU hay không bằng cách sử dụng `torch.backends.mps.is_available()`.

````python
# Check for Apple Silicon GPU
import torch
torch.backends.mps.is_available() # Note this will print false if you're not running on a Mac
````

````python
# Set device type
device = "mps" if torch.backends.mps.is_available() else "cpu"
device
````

Tương tự như trước, nếu mã ở trên trả về `"mps"`, điều đó có nghĩa là chúng ta có thể thiết lập toàn bộ mã PyTorch của mình để sử dụng GPU Apple Silicon có sẵn.

````python
if torch.cuda.is_available():
    device = "cuda" # Use NVIDIA GPU (if available)
elif torch.backends.mps.is_available():
    device = "mps" # Use Apple Silicon GPU (if available)
else:
    device = "cpu" # Default to CPU if no GPU is available
````

### 3. Đưa tensors (và các mô hình) vào GPU

Bạn có thể đưa tensors (và các mô hình, chúng ta sẽ thấy ở phần sau) vào một thiết bị cụ thể bằng cách gọi [`to(device)`](https://pytorch.org/docs/stable/generated/torch.Tensor.to.html) trên chúng. Trong đó `device` là thiết bị mục tiêu mà bạn muốn đưa tensor (hoặc mô hình) vào.

Tại sao phải làm việc này?

GPU cung cấp tốc độ tính toán số học nhanh hơn nhiều so với CPU và nếu không có GPU, nhờ vào mã không phụ thuộc thiết bị (device agnostic code - xem ở trên) của chúng ta, nó sẽ chạy trên CPU.

> **Lưu ý:** Đưa một tensor vào GPU bằng `to(device)` (ví dụ: `some_tensor.to(device)`) sẽ trả về một bản sao của tensor đó, ví dụ: cùng một tensor sẽ nằm trên cả CPU và GPU. Để ghi đè lên các tensors, hãy gán lại chúng:
>
> `some_tensor = some_tensor.to(device)`

Hãy thử tạo một tensor và đưa nó vào GPU (nếu có sẵn).

````python
# Create tensor (default on CPU)
tensor = torch.tensor([1, 2, 3])

# Tensor not on GPU
print(tensor, tensor.device)

# Move tensor to GPU (if available)
tensor_on_gpu = tensor.to(device)
tensor_on_gpu
````

Nếu bạn có một GPU khả dụng, đoạn mã trên sẽ xuất ra thứ gì đó giống như:

```
tensor([1, 2, 3]) cpu
tensor([1, 2, 3], device='cuda:0')
```

Lưu ý rằng tensor thứ hai có `device='cuda:0'`, điều này có nghĩa là nó được lưu trữ trên GPU thứ 0 có sẵn (các GPU được lập chỉ mục từ 0, nếu có hai GPU có sẵn, chúng sẽ lần lượt là `'cuda:0'` và `'cuda:1'`, lên tới `'cuda:n'`).

### 4. Đưa tensors trở lại CPU

Điều gì sẽ xảy ra nếu chúng ta muốn đưa tensor trở lại CPU?

Ví dụ: bạn sẽ muốn làm điều này nếu bạn muốn tương tác với các tensors của mình bằng NumPy (NumPy không tận dụng được GPU).

Hãy thử sử dụng phương thức [`torch.Tensor.numpy()`](https://pytorch.org/docs/stable/generated/torch.Tensor.numpy.html) trên biến `tensor_on_gpu` của chúng ta.

````python
# If tensor is on GPU, can't transform it to NumPy (this will error)
tensor_on_gpu.numpy()
````

Thay vào đó, để đưa một tensor về lại CPU và có thể sử dụng được với NumPy, chúng ta có thể sử dụng [`Tensor.cpu()`](https://pytorch.org/docs/stable/generated/torch.Tensor.cpu.html).

Thao tác này sẽ sao chép tensor vào bộ nhớ của CPU để nó có thể được sử dụng với các CPU.

````python
# Instead, copy the tensor back to cpu
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
````

Đoạn mã trên trả về một bản sao của tensor GPU trong bộ nhớ CPU, do đó tensor ban đầu vẫn nằm trên GPU.

````python
tensor_on_gpu
````
